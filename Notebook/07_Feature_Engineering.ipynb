## <U>Etapa 1</U>: Importacion de las librerias y cargar el archivo a analizar

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

print("Libreria importadas correctamente")
```

### Cargar archivo a analizar

```python
df_perfiles = pd.read_csv(
    "../Datos/Sinteticos/perfiles_estudiantes_sinteticos.csv",
    encoding="utf-8-sig"
)

print("Dimensiones:", df_perfiles.shape)

display(df_perfiles.head())
```

## <U>Etapa 2:</U> Seleccionar las variables para el analisis

### Variables de intereses

```python
variables_intereses = [
    "interes_salud",
    "interes_tecnologia",
    "interes_educacion",
    "interes_arte",
    "interes_social",
    "interes_ambiente"
]
print("Diccionario cargado correctamente")
```

### Variables de fortalezas

```python
variables_fortalezas = [
    "fortaleza_matematica",
    "fortaleza_lengua",
    "fortaleza_biologia",
    "fortaleza_historia",
    "fortaleza_fisica",
    "fortaleza_tecnologia"
]
print("Diccionario cargado correctamente")
```

### Unificar las variables 

```python
variables_modelo = variables_intereses + variables_fortalezas
print("Variables unificadas correctamente")
```

### Crear un Dataframe con las variables seleccionadas

```python
X = df_perfiles[variables_modelo].copy()

print("Cantidad de perfiles:", X.shape[0])
print("Cantidad de variables:", X.shape[1])

display(X.head())
```

## <U>Etapa 3</U>: Dividir el Dataset

### Separar los datos de entrenamiento y prueba

```python
X_train, X_test = train_test_split(X,test_size=0.20,random_state=42,shuffle=True)

print("Perfiles para entrenamiento:", X_train.shape[0])
print("Perfiles para prueba:", X_test.shape[0])

print("\nTotal de perfiles:", X_train.shape[0] + X_test.shape[0])
```

## <U>Etapa 4</U>: Normalizar las variables
### Ahora aplicaremos **"MinMaxScaler"** para transformar las 12 variables a una escala entre 0 y 1.

### Crear escalador

```python
scaler = MinMaxScaler()
print("Escalador creado correctamente")
```

### Ajustar con entrenamiento y transformar

```python
X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train),
    columns=variables_modelo,
    index=X_train.index
)
print("Ajustado")
```

### Transformar prueba utilizando el mismo escalador

```python
X_test_scaled = pd.DataFrame(
    scaler.transform(X_test),
    columns=variables_modelo,
    index=X_test.index
)

print("Dimensiones de entrenamiento:", X_train_scaled.shape)
print("Dimensiones de prueba:", X_test_scaled.shape)

print("\nValores mínimos y máximos del entrenamiento:")
display(X_train_scaled.agg(["min", "max"]).T)

print("\nPrimeros perfiles normalizados:")
display(X_train_scaled.head().round(2))
```

## <U>Etapa 5</U>: Guardar los datos preparados

### Utilizaremos Joblib para guardar el escalador para usarlo en otro momento

```python
import joblib
print("Importado correctamente")
```

### Guardar los conjuntos normalizados

```python
X_train_scaled.to_csv(
    "../Datos/Procesados/perfiles_entrenamiento.csv",
    index=False
)

X_test_scaled.to_csv(
    "../Datos/Procesados/perfiles_prueba.csv",
    index=False
)
print("guardado correctamente")
```

### Guardar los identificadores originales por separado

```python
ids_train = df_perfiles.loc[X_train_scaled.index, ["perfil_id"]]
ids_test = df_perfiles.loc[X_test_scaled.index, ["perfil_id"]]

ids_train.to_csv(
    "../Datos/Procesados/ids_entrenamiento.csv",
    index=False
)

ids_test.to_csv(
    "../Datos/Procesados/ids_prueba.csv",
    index=False
)
print("Guardado correctamente")
```

### Guardar el escalador para utilizarlo mas adelante

```python
joblib.dump(
    scaler,
    "../Datos/Procesados/escalador_perfiles.pkl"
)

print("Archivos guardados correctamente.")
```
