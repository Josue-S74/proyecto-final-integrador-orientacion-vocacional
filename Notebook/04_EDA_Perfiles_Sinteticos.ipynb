## <U>Etapa 2</U>: Visualizacion de los datos

## Histogramas

### Fortalezas

```python
fortalezas = [
    "fortaleza_matematica",
    "fortaleza_lengua",
    "fortaleza_biologia",
    "fortaleza_historia",
    "fortaleza_fisica",
    "fortaleza_tecnologia"
]
```

```python
fig, axes = plt.subplots(2,3, figsize =(15,8))

for i, columna in enumerate(fortalezas):
    sns.histplot(
        df[columna],
        bins=10,
        kde=True,
        ax=axes[i//3,i%3]
    )
    
    axes[i//3,i%3].set_title(columna)
    axes[i//3,i%3].set_xlabel("Calificacion")
    axes[i//3,i%3].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

### Intereses

```python
intereses= [
    "interes_salud",
    "interes_tecnologia",
    "interes_educacion",
    "interes_arte",
    "interes_social",
    "interes_ambiente"
]
```

```python
fig, axes = plt.subplots(2,3, figsize =(15,8))

for i, columna in enumerate(intereses):
    sns.histplot(
        df[columna],
        bins=5,
        kde=False,
        ax=axes[i//3,i%3]
    )
    
    axes[i//3,i%3].set_title(columna)
    axes[i//3,i%3].set_xlabel("Nivel de interes")
    axes[i//3,i%3].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

### Preferencias de trabajo

```python
preferencias =[
    "preferencia_trabajo_equipo",
    "preferencia_trabajo_practico",
    "preferencia_trabajo_autonomo",
    "preferencia_investigacion"
]
```

```python
fig, axes = plt.subplots(2,2, figsize =(12,8))

for i, columna in enumerate(preferencias):
    sns.histplot(
        df[columna],
        bins=5,
        kde=False,
        ax=axes[i//2,i%2]
    )
    
    axes[i//2,i%2].set_title(columna)
    axes[i//2,i%2].set_xlabel("Nivel de preferencias")
    axes[i//2,i%2].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

## Boxplots

### Fortalezas

```python
plt.figure(figsize=(12,6))

sns.boxplot(data=df[fortalezas])

plt.title("Distribucion de las fortalezas academicas")
plt.xlabel("Areas academicas")
plt.ylabel("Calificacion")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### Intereses

```python
plt.figure(figsize=(12,6))

sns.boxplot(data=df[intereses])

plt.title("Distribucion de las intereses personales")
plt.xlabel("Areas interes")
plt.ylabel("Nivel de interes")

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### Preferencias de trabajo

```python
plt.figure(figsize=(12,6))

sns.boxplot(data=df[preferencias])

plt.title("Distribucion de las preferencias")
plt.xlabel("Preferencias")
plt.ylabel("Nivel de preferencias")

plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

## Matriz de Correlacion


### Seleccionar las variables

```python
variables_numericas = fortalezas + intereses + preferencias
print("Variables cargadas correctamente")
```

### Calcular correlaciones

```python
correlaciones = df[variables_numericas].corr()

correlaciones.round(2)
```

### Mapa de calor


```python
plt.figure(figsize=(12,9))

sns.heatmap(
    correlaciones,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0,
    annot_kws={"size":7}
)

plt.xticks(rotation=40)
plt.title("Matriz de correlacion de los perfiles sinteticos")
plt.tight_layout()
plt.show()
```

### Relacion entre fortalezas e intereses

```python
correlacion_cruzada = df[fortalezas+intereses].corr()

correlacion_cruzada = correlacion_cruzada.loc[fortalezas,intereses]

plt.figure(figsize=(11,6))

sns.heatmap(
    correlacion_cruzada,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0,
)

plt.xticks(rotation=40)
plt.title("Correlacion entre fortalezas e intereses")
plt.tight_layout()
plt.show()
```

# <u>Concluciones del EDA de perfiles sinteticos

### El analisis exploratorio permitio comprender la distribucion y el comportamiento de las variables que componen los 1000 perfiles estudiantiles sinteticos.
### Se observaron diferencias entre las fortalezas academicas, los intereses personales y las preferencias de trabajo. Las correlaciones entre fortalezas e intereses fueron muy debiles, lo que demuestra que ambas dimensiones deben de analizarse por separado y combinarse cuidadosamente en el sistema de orientacion vocacional.
### Tambien se comprobo que los perfiles presentan distintas combinaciones de caracteristicas, permitiendo representar multiples preferencias y fortalezas.
### <U>Limitaciones:</U> los datos son ficticios y fueron generados artificialmente. Por lo tanto, sus distribuciones y relaciones no representan necesariamente el comportamiento de estudiantes reales. Los resultados sirven para probar el funcionamiento tecnico del prototipo, pero no permiten validar cientificamente sus recomendaciones.
### <U>Decisiones para el desarrollo:</U> el recomendador debera considerar tanto las fortalezas academicas como los intereses personales. Las preferencias de trabajo podran utilizarse como una dimension complementaria. La ubicacion geografica se utilizara para informar sobre la oferta educativa disponible, sin modificar artificialmente la afinidad personal del estudiante
