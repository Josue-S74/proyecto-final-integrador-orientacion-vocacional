## <U>Etapa 4</U>: Creacion del DataFrame de trabajo

```python
df_limpio=df.copy()
print("DataFrame copiado exitosamente")
```

## <U>Etapa 5</U>: Eliminacion de duplicados exactos

```python
duplicados_antes = df_limpio.duplicated().sum()

df_limpio=df_limpio.drop_duplicates(keep="first").copy()

duplicados_despues = df_limpio.duplicated().sum()

print("Duplicados antes: ",duplicados_antes)
print("Duplicados eliminados: ",duplicados_antes-duplicados_despues)
print("Duplicados restantes: ",duplicados_despues)
print("Tamaño final: ",df_limpio.shape)
```

## <U>Etapa 6</U>: Normalizacion de localidades

```python
df_limpio["Localidad_normalizada"] = df_limpio["Localidad"].copy()
print("Se creo correctamente la nueva columna")
```

### Creacion de diccionario para corregir localidades

```python
correcion_localidad = {
    "San Salvado de Jujuy":"San Salvador de Jujuy",
    "Palpala":"Palpalá",
    "Maimara":"Maimará",
    "Libertador General San Martin":"Libertador General San Martín",
    "Libertador Gral. San Martín":"Libertador General San Martín"
}
```

### Normalizando las localidades

```python
df_limpio["Localidad_normalizada"] = (df_limpio["Localidad_normalizada"].replace(correcion_localidad))
```

```python
print("Localidades originales: ",df_limpio["Localidad"].nunique())

print("Localidades normalizadas: ",df_limpio["Localidad_normalizada"].nunique())
```

### Comparar localidades (originales y normalizadas)

```python
comparacion_localidades = df_limpio[["Localidad","Localidad_normalizada"]].drop_duplicates()
comparacion_localidades.sort_values(by="Localidad")
```

### Mostrar unicamente las localidades modificadas

```python
localidades_modificadas = df_limpio[df_limpio["Localidad"]!=df_limpio["Localidad_normalizada"]]
localidades_modificadas[["Localidad","Localidad_normalizada"]].drop_duplicates()
```

### Comprobar la frecuencia despues de normalizar

```python
df_limpio["Localidad_normalizada"].value_counts()
```
